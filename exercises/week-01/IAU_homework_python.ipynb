You are given an array of strings:

```python
from statsmodels.tsa.statespace import news

array_to_split = ["Fiit", "fiit", "megadlhystring!!!", "autobiografia", "Alica", "python", "pip", "env", "jupyter", "IAU"]
```

Write a function that, for the given array of strings, returns a new array containing only strings at prime indices. Then call this function with the field declared above as a parameter.

```python
def is_prime(n):
    if n<2:
        return False
    if n == 2 or n == 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    k = 5
    while k * k <= n:
        if n % k == 0 or n % (k + 2) == 0:
            return False
        k += 6
    return True


def get_prime_array(array):
    new = []
    for i in range(len(array)):
        if is_prime(i):
            new.append(array[i])
    return new

get_prime_array(array_to_split)
```

Write a program that divides the above declared array into 3 new arrays, where in one array there will be strings of length less than 5. In the second array, strings of length between 5 and 10 (inclusive) and strings starting with the letter "a" or any uppercase letter while string of **any** length. In the third field, all other thongs.

```python
def divide(array):
    first = []
    second = []
    third = []

    for element in array:
        if len(element) < 5:
            first.append(element)
        elif len(element) <= 10 or element[0] =='a' or element[0].isupper():
            second.append(element)
        else:
            third.append(element)
    return first, second, third

first, second, third = divide(array_to_split)
print(first)
print(second)
print(third)


```

Write a function that tells us, for an array of ints, whether the array can be split into two arrays with the same sum. Then call this function on the field declared above. List two such fields (any that satisfy the condition of equal sum).

```python
def find_half(array, target, i, chosen):
    if target == 0:
        return chosen
    if i == len(array) or target < 0:
        return None
    return find_half(array, target - array[i], i + 1, chosen + [i]) or find_half(array, target, i + 1, chosen)

def split_equal(array):
    if sum(array) % 2 != 0:
        return None
    chosen = find_half(array, sum(array) // 2, 0, [])
    if chosen is None:
        return None
    first, second = [], []
    for i in range(len(array)):
        if i in chosen:
            first.append(array[i])
        else:
            second.append(array[i])
    return first, second

result = split_equal(array)
if result is None:
    print("none")
else:
    print(result[0], sum(result[0]))
    print(result[1], sum(result[1]))
```
